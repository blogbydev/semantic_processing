```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data
documents = [
    "I love programming in Python",
    "Python is great for machine learning",
    "Machine learning is fascinating"
]
```

```python
# Create an instance of CountVectorizer
vectorizer = CountVectorizer(stop_words='english')
```

```python
# Fit and transform the text data
X = vectorizer.fit_transform(documents)
```

```python
# Get the feature names (words in the vocabulary)
feature_names = vectorizer.get_feature_names_out()

# Convert the matrix to an array for better readability
print("Feature Names:", feature_names)
print("Document-Term Matrix:\n", X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data
documents = ["I love programming in Python"]

# Unigram (default case, single words)
vectorizer_unigram = CountVectorizer(ngram_range=(1, 1))
X_unigram = vectorizer_unigram.fit_transform(documents)
print("Unigram Features:", vectorizer_unigram.get_feature_names_out())
print(X_unigram.toarray())

# Bigram (pairs of words)
vectorizer_bigram = CountVectorizer(ngram_range=(2, 2))
X_bigram = vectorizer_bigram.fit_transform(documents)
print("\nBigram Features:", vectorizer_bigram.get_feature_names_out())
print(X_bigram.toarray())

# Trigram (triplets of words)
vectorizer_trigram = CountVectorizer(ngram_range=(3, 3))
X_trigram = vectorizer_trigram.fit_transform(documents)
print("\nTrigram Features:", vectorizer_trigram.get_feature_names_out())
print(X_trigram.toarray())

# Mixed range (unigram + bigram)
vectorizer_mixed = CountVectorizer(ngram_range=(1, 2))
X_mixed = vectorizer_mixed.fit_transform(documents)
print("\nMixed Unigram + Bigram Features:", vectorizer_mixed.get_feature_names_out())
print(X_mixed.toarray())
```
